```python
# Import required libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

%matplotlib inline

print("Libraries imported successfully")
```

```python
# Create student dataset

data = {
    "Name": ["Aman", "Riya", "Rahul", "Simran", "Karan", "Neha", "Arjun", "Priya"],
    "Maths": [78, 90, 65, 88, 72, 95, 60, 85],
    "Science": [85, 88, 70, 92, 75, 89, 65, 80],
    "English": [82, 91, 60, 85, 70, 94, 58, 87]
}

df = pd.DataFrame(data)

df
```

```python
# Calculate descriptive statistics

df.describe()
```

```python
# Histogram visualization

df.hist(figsize=(8,6))

plt.suptitle("Distribution of Student Marks")

plt.show()
```

```python
# Calculate descriptive statistics

mean = df.mean(numeric_only=True)
median = df.median(numeric_only=True)
mode = df.mode(numeric_only=True).iloc[0]
std = df.std(numeric_only=True)

print("Mean:\n", mean)
print("\nMedian:\n", median)
print("\nMode:\n", mode)
print("\nStandard Deviation:\n", std)

```

```python
# Bar chart of average marks

mean.plot(kind='bar')

plt.title("Average Marks per Subject")
plt.xlabel("Subjects")
plt.ylabel("Average Marks")

plt.show()
```

## Interpretation and Insights

The descriptive statistics show that the average marks of students in Maths, Science, and English are around 75–85, indicating good academic performance.

The histogram shows that most students scored between 60 and 95, with no extreme outliers.

The bar chart shows that Science has the highest average marks, followed by Maths and English.

## Conclusion

Overall, student performance is consistent and balanced across all subjects.
