# 📊 Week 1 Assignment: Student Marks Data Analysis

This notebook demonstrates basic data analysis using Python, Pandas, and Matplotlib.

Objectives:
- Load dataset from CSV file
- Perform descriptive statistical analysis
- Visualize score distribution using histogram
- Interpret insights from data

```python
# Import Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
#Load dataset
df = pd.read_csv("student_marks.csv")
df.head()

df_display = df.copy()

# Start index from 0
df_display.index = df_display.index+1

# Reset index so S.No becomes a proper column (fixes blank row issue)
df_display = df_display.reset_index()

# Rename index column to S.No
df_display = df_display.rename(columns={"index": "S.No"})

# Display professional table with full borders
df_display.style \
    .set_caption("Student Marks Dataset") \
    .hide(axis="index") \
    .set_table_styles([
        {
            'selector': 'table',
            'props': [
                ('border', '2px solid black'),
                ('border-collapse', 'collapse'),
                ('width', '70%'),
                ('margin', '20px auto'),
                ('font-size', '14px')
            ]
        },
        {
            'selector': 'th',
            'props': [
                ('border', '1px solid black'),
                ('padding', '8px'),
                ('text-align', 'center'),
                ('background-color', '#e6e6e6'),
                ('font-weight', 'bold')
            ]
        },
        {
            'selector': 'td',
            'props': [
                ('border', '1px solid black'),
                ('padding', '8px'),
                ('text-align', 'center')
            ]
        },
        {
            'selector': 'caption',
            'props': [
                ('caption-side', 'top'),
                ('font-size', '18px'),
                ('font-weight', 'bold'),
                ('padding', '10px')
            ]
        }
    ])
```

```python
## Descriptive Statistics

# Select numeric columns
numeric_df = df[['Maths', 'Science', 'English']]

# Create professional descriptive statistics table
desc_stats = pd.DataFrame({
    "Subject": numeric_df.columns,
    "Mean": numeric_df.mean().values,
    "Median": numeric_df.median().values,
    "Mode": numeric_df.mode().iloc[0].values,
    "Standard Deviation": numeric_df.std().values
})

# Apply professional styling with borders and colors
desc_stats.style \
    .set_caption("Descriptive Statistical Summary of Student Marks") \
    .hide(axis="index") \
    .format({
        "Mean": "{:.2f}",
        "Median": "{:.2f}",
        "Mode": "{:.0f}",
        "Standard Deviation": "{:.2f}"
    }) \
    .set_table_styles([
        {'selector': 'table', 'props': [
            ('border', '2px solid black'),
            ('border-collapse', 'collapse'),
            ('width', '70%'),
            ('text-align', 'center')
        ]},
        {'selector': 'th', 'props': [
            ('border', '1px solid black'),
            ('padding', '10px'),
            ('background-color', '#d9ead3'),
            ('font-weight', 'bold')
        ]},
        {'selector': 'td', 'props': [
            ('border', '1px solid black'),
            ('padding', '8px')
        ]},
        {'selector': 'caption', 'props': [
            ('font-size', '18px'),
            ('font-weight', 'bold'),
            ('padding', '10px')
        ]}
    ])
```

## 📊 Visualization of Student Marks Distribution

This section uses histograms and a bar graph to visualize student marks and support statistical analysis.

### Histogram
- Shows the distribution and spread of marks across Maths, Science, and English  
- Helps understand score frequency and variation  
- Supports analysis of overall performance patterns  

### Bar Graph
- Displays the average marks for each subject  
- Helps compare overall performance between Maths, Science, and English  
- Clearly highlights differences in subject-wise performance  


```python
import matplotlib.pyplot as plt
import numpy as np

bins = [50, 60, 70, 80, 90, 100]

plt.figure(figsize=(10,6))

plt.hist([df['Maths'], df['Science'], df['English']],
         bins=bins,
         label=['Maths', 'Science', 'English'],
         edgecolor='black')

plt.title("Histogram of Student Marks Distribution by Subject",
          fontsize=16, fontweight='bold')

plt.xlabel("Marks Range", fontsize=12)
plt.ylabel("Number of Students", fontsize=12)

plt.xticks(bins)

plt.legend(title="Subjects")

plt.grid(axis='y', linestyle='--', alpha=0.6)

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset
df = pd.read_csv("student_marks.csv")

# Calculate average marks
subjects = ["Maths", "Science", "English"]
subject_avg = [
    df["Maths"].mean(),
    df["Science"].mean(),
    df["English"].mean()
]

# Define different colors for each subject
colors = ["skyblue", "lightgreen", "orange"]

# Create bar chart
plt.figure(figsize=(8,6))
bars = plt.bar(subjects, subject_avg, color=colors, edgecolor="black")

# Add value labels on top of bars
for bar in bars:
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2, height + 0.5,
             f'{height:.1f}', ha='center', fontsize=11)

# Labels and title
plt.title("Average Marks by Subject", fontsize=14)
plt.xlabel("Subjects", fontsize=12)
plt.ylabel("Average Marks", fontsize=12)

# Add grid for better readability
plt.grid(axis="y", linestyle="--", alpha=0.7)

# Show chart
plt.show()
```

## Interpretation of Insights

- Most student scores are concentrated within a consistent range.
- The mean and median values are close, indicating balanced score distribution.
- The score spread shows moderate variation across subjects.
- Visualizations confirm stable and uniform performance.
- No significant outliers were detected in the dataset.
- Overall, the dataset shows consistent academic performance.
